# Generate Degeneracy

```python
def generate_classical_degeneracy_data(length, num_ones, seedlist):
    seeddict = {}
    for seed in seedlist:
        jij = generate_jij(length, num_ones, seed)
        Hint, Hintdiag = sg_interaction(length, jij)
        minval = min(Hintdiag)
        minenergy = minval.real
        degeneracy = np.count_nonzero(Hintdiag == minenergy)
        seeddict.update({seed:degeneracy})
    #sorted_seeds = {k: seeddict[k] for k in sorted(seeddict, key=lambda k: seeddict[k], reverse=True)}
    return seeddict #sorted_seeds
```

```python
generate_classical_degeneracy_data(8, 14, [1,2,3])
```

# Generate Classical Ground States

```python
def generate_classical_gs_data(length, num_ones, seedlist):
    seeddict = {}
    for seed in seedlist:
        jij = generate_jij(length, num_ones, seed)
        Hint, Hintdiag = sg_interaction(length, jij)
        Hintdiag, minenergy, degeneracy, groundpos, states = find_degeneracy(length, Hintdiag)
        seeddict.update({seed:states})
    return seeddict
```

```python
generate_classical_gs_data(8, 14, [1,2,3])
```

# Generate Quantum Groundstate

```python
def generate_psi_data(length, num_ones, seedlist, Hfield, hval):
    seeddict = {}
    for seed in seedlist:
        jij = generate_jij(length, num_ones, seed)
        Hint, Hintdiag = sg_interaction(length, jij)
        H = -1 * (Hint + (hval * Hfield))
        E, psi = H.groundstate()
        seeddict.update({seed:psi})
    return seeddict
```

```python
Hfield = sg_field(8)
generate_psi_data(8, 14, [1,2,3], Hfield, 0.001)
```

# Generate Mutual Information

```python
def generate_mutual_information_data(length, num_ones, seedlist, Hfield, hval):
    seeddict = {}
    for seed in seedlist:
        jij = generate_jij(length, num_ones, seed)
        Hint, Hintdiag = sg_interaction(length, jij)
        H = -1 * (Hint + (hval * Hfield))
        E, psi = H.groundstate()
        minfo1 = []
        for i in range(length):
            for j in range(i + 1, length):
                combinations, physdist, mutualinfo, distance= mutual_info(length, psi, [i], [j])
                minfo1.append(mutualinfo)
        seeddict.update({seed:minfo1})
    return seeddict
```
